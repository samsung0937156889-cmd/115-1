```python
s = 7
guess = int(input("請猜一個數: "))
while guess != s:
    if guess > s:
        print("你猜太大了")
    else:
        print("你猜太小了")
    guess = int(input("請再猜一個數: "))
print("恭喜猜中了")
```

```python
import random
day=1
max_day=3
win=False
while day<=max_day and not win:
  lucky=random.randint(1,50)
  guess=int(input(f"今天第{day}天,請輸入1-50數字: "))
  print(f"今天的幸運數字是{lucky}")
  if guess == lucky:
    print("你中獎了")
    win=True
  else:
    print("你沒中獎，明天再來")
  day+=1
if not win:
  print("很遺憾3天都沒中獎，下次再來試試吧")
```

```python
import random

while True:
    event = random.randint(1, 5)

    if event == 1:
        print("遇到怪物")
    elif event == 2:
        print("找到寶物")
    elif event == 3:
        # 陷阱跳過
        continue
    elif event == 4:
        print("補充生命")
    elif event == 5:
        print("出口結束")
        break
```
